```python
etherpad_url = "http://localhost:9001/"
migration_pad_title = "Solr Migration Test Pad"
default_result_path = None
close_on_fail = False
transition_timeout = 10000
```

# Solr Volume Migration Test - Phase 2: The new Solr with the old volume

The old volume contains the old core configuration, which the new Solr cannot load.
Etherpad itself must stay up, but the pad must no longer be searchable.

```python
import tempfile

work_dir = tempfile.mkdtemp()
if default_result_path is None:
    default_result_path = work_dir
work_dir
```

```python
import importlib

import scripts.playwright
importlib.reload(scripts.playwright)

from scripts.playwright import *

await init_pw_context(close_on_fail=close_on_fail, last_path=default_result_path)
```

## Etherpad is up, but searching fails.

```python
async def _step(page):
    await page.goto(etherpad_url)

    await expect(page.locator(".hashview-search-box")).to_be_editable()
    await expect(page.locator(f'text="{migration_pad_title}"')).to_have_count(0, timeout=transition_timeout)

await run_pw(_step)
```

Clean up

```python
await finish_pw_context()
```
